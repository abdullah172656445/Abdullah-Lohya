This notebook answers all five questions using the supplied `student_tea_snacks.csv` dataset.


## 1. Import libraries and load the dataset


```python
import pandas as pd
import matplotlib.pyplot as plt

# Keep the CSV in the same folder as this notebook.
df = pd.read_csv('student_tea_snacks.csv')
df.head()
```

```python
# Check the dataset structure and missing values
print('Dataset shape:', df.shape)
print('\nMissing values:')
print(df.isna().sum())
print('\nTea preference responses:')
print(df['prefers_tea'].value_counts())
```

## Q1. What is the most preferred tea type?

Only students who answered **Yes** to `prefers_tea` are included. Students who do not drink tea have no tea type recorded, so they should not be counted as a tea choice.


```python
tea_drinkers = df[df['prefers_tea'].str.strip().str.lower() == 'yes']
tea_type_counts = tea_drinkers['tea_type'].value_counts()
print(tea_type_counts)
print('\nMost preferred tea type:', tea_type_counts.idxmax(),
      '(', tea_type_counts.max(), 'students)')
```

## Q2. What is the average number of cups per gender?


```python
average_cups_by_gender = df.groupby('gender')['cups_per_day'].mean()
print(average_cups_by_gender.round(2))
```

## Q3. Make a bar chart of snack choices


```python
snack_counts = df['snack_choice'].value_counts()
print(snack_counts)
snack_counts.plot(kind='bar', figsize=(8, 5))
plt.title('Student Snack Choices')
plt.xlabel('Snack Choice')
plt.ylabel('Number of Students')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Q4. Compare tea preference by age group

The comparison below includes tea drinkers only and shows both counts and percentages within each age group.


```python
tea_by_age = pd.crosstab(tea_drinkers['age_group'], tea_drinkers['tea_type'])
print('Counts:')
display(tea_by_age)
print('Percentages within each age group:')
display((tea_by_age.div(tea_by_age.sum(axis=1), axis=0) * 100).round(1))
```

## Q5. Count the number of students who don't drink tea


```python
non_tea_drinkers = df[df['prefers_tea'].str.strip().str.lower() == 'no']
print('Number of students who do not drink tea:', len(non_tea_drinkers))
```

## Summary of answers

- **Q1:** Green tea is the most preferred type among students who drink tea and have a recorded tea type (17 students).
- **Q2:** Average cups per day: Female = 2.60; Male = 3.00.
- **Q3:** Samosa is the most selected snack (40 students), followed by Chips (31) and Biscuit (29).
- **Q4:** Among tea drinkers with a recorded tea type, Green tea is most common in age groups 18–20 (8), and 24–26 (6). Black tea is most common in age group 21–23 (4).
- **Q5:** 49 students do not drink tea.

**Data-quality note:** The dataset has 51 students marked as tea drinkers, but 12 of them have no tea type recorded. Also, 38 students marked as non-tea-drinkers have a tea type recorded. Q1 and Q4 therefore use the recorded tea types for students marked Yes, and exclude missing tea types.
