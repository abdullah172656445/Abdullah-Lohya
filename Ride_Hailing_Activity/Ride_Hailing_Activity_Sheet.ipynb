Completed activity sheet using the supplied 100-trip CSV dataset. Run cells from top to bottom.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("ride_hailing_dataset.csv")
display(df.head())
print("Dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
```

## Q1. What is the average fare by service type?

```python
average_fare = df.groupby("service_type")["fare"].mean().sort_values(ascending=False)
display(average_fare.round(2).rename("Average fare (PKR)").to_frame())
```

**Answer:** InDrive has the highest average fare at **PKR 573.51**. Careem averages **PKR 544.98**, Rickshaw **PKR 486.64**, and Uber **PKR 486.13**.

## Q2. Which city has the most trips?

```python
trips_by_city = df["city"].value_counts()
display(trips_by_city.rename("Number of trips").to_frame())
print(f"{trips_by_city.idxmax()} has the most trips, with {trips_by_city.max()} trips.")
```

**Answer:** Islamabad has the most trips, with **29 trips**.

## Q3. What is the average distance per trip?

```python
average_distance = df["distance_km"].mean()
print(f"Average distance per trip: {average_distance:.2f} km")
```

**Answer:** The average distance per trip is **9.61 km**.

## Q4. Compare trip ratings by gender.

```python
ratings_by_gender = df.groupby("gender")["trip_rating"].agg(["count", "mean", "median"]).round(2)
display(ratings_by_gender)
```

**Answer:** Male passengers have a higher average rating (**3.33**) than female passengers (**2.82**), a difference of approximately **0.50 points**. There are 49 male trips and 51 female trips.

## Q5. Make a bar chart of service type vs. average fare.

```python
average_fare = df.groupby("service_type")["fare"].mean().sort_values(ascending=False)
plt.figure(figsize=(8, 5))
average_fare.plot(kind="bar")
plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare (PKR)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Conclusion

The dataset contains 100 trips and no missing values. InDrive has the highest average fare, Islamabad has the most trips, and the average trip distance is 9.61 km. Male passengers have a higher average trip rating than female passengers in this dataset.
