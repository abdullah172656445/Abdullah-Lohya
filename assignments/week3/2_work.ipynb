# Assignment 3 — Two Averages, One Argument
**The Prices Desk | Week 3 — Descriptive analytics**

This notebook uses the actual `teaching_prices_messy.csv` uploaded for this assignment and reapplies the cleaning decisions documented in Assignment 2. It preserves the raw price, removes one exact duplicate only from the analytical table, parses comma-formatted numbers, and retains uncertain observations with flags.

**Data label:** illustrative teaching data — not actual PBS observations.


```python
import pandas as pd
import numpy as np

raw = pd.read_csv("teaching_prices_messy.csv", dtype={"price": "string"})
df = raw.copy()
df["price_raw"] = df["price"]
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df["item_clean"] = (df["item"].astype("string").str.strip()
                    .str.replace(r"\\s+", " ", regex=True).str.title())
df["price_num"] = pd.to_numeric(
    df["price_raw"].str.replace(",", "", regex=False), errors="coerce"
)
df["exact_duplicate"] = df.duplicated(keep="first")
df["missing_price"] = df["price_num"].isna()
df["extreme_9999"] = df["price_num"].eq(9999)
df["suspicious_oil_city1_low"] = (
    df["item_clean"].eq("Oil B") & df["city"].eq("City 1") & df["price_num"].lt(200)
)
df["comma_formatted"] = df["price_raw"].str.contains(",", na=False)
df["review_flag"] = df[["missing_price", "extreme_9999",
                        "suspicious_oil_city1_low"]].any(axis=1)

audit = df.copy()
clean = df.loc[~df["exact_duplicate"]].copy()
print("Raw rows:", len(raw), "| Analytical rows:", len(clean))
print("Missing prices:", clean["price_num"].isna().sum())
print("Flagged 9999:", clean["extreme_9999"].sum())
print("Suspicious low Oil B / City 1:", clean["suspicious_oil_city1_low"].sum())
print("Comma-formatted parsed values:", clean.loc[clean["comma_formatted"], "price_num"].tolist())
```

## 1. Mean, median and first-to-last percentage change

Mean and median use all available numeric prices for each item in the cleaned analytical table. Missing prices are omitted; flagged numeric values remain included.

For the first-to-last change, calculate the mean of the available city prices for that item on its first date and last date, then:

\[
\\text{Change (\\%)}=\\frac{\\text{last-date mean}-\\text{first-date mean}}{\\text{first-date mean}}\\times100
\]


```python
summary_rows = []
for item, group in clean.groupby("item_clean", sort=True):
    values = group["price_num"].dropna()
    date_means = group.groupby("date")["price_num"].mean().sort_index()
    first_mean, last_mean = date_means.iloc[0], date_means.iloc[-1]
    summary_rows.append({
        "Item": item, "Numeric observations": len(values),
        "Mean (PKR/kg)": values.mean(), "Median (PKR/kg)": values.median(),
        "First-date mean": first_mean, "Last-date mean": last_mean,
        "First-to-last change (%)": (last_mean-first_mean)/first_mean*100
    })
summary = pd.DataFrame(summary_rows)
display(summary.round(2))
summary.to_csv("item_summary.csv", index=False)
```

## 2. One flagged value: Sweetener D with and without 9999

The 9999 value remains in the main cleaned data because its intended correction is unknown. Excluding it below is only a sensitivity comparison, not a correction.


```python
sweetener = clean.loc[clean["item_clean"].eq("Sweetener D"), "price_num"].dropna()
without_9999 = sweetener.loc[sweetener.ne(9999)]
sensitivity = pd.DataFrame([
    {"Calculation":"Including flagged 9999","n":len(sweetener),
     "Mean":sweetener.mean(),"Median":sweetener.median()},
    {"Calculation":"Excluding flagged 9999 for sensitivity check","n":len(without_9999),
     "Mean":without_9999.mean(),"Median":without_9999.median()}
])
display(sensitivity.round(2))
sensitivity.to_csv("sweetener_sensitivity.csv", index=False)
```

## 3. Review flagged rows and export files


```python
display(clean.loc[
    clean["review_flag"] | clean["comma_formatted"],
    ["date","item","item_clean","city","price_raw","price_num",
     "missing_price","extreme_9999","suspicious_oil_city1_low","comma_formatted"]
].sort_values(["date","item_clean","city"]))
clean.to_csv("teaching_prices_cleaned_with_flags.csv", index=False)
audit.to_csv("teaching_prices_audit_all_rows.csv", index=False)
print("Exports saved.")
```
